```python
import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# only district 2 shape
# you can filter the country
shp_district2 = gpd.read_file("merged_adm2.shp")
shp_district2.head()
```

```python
#plot
target_crs = "EPSG:4326"
shp_district2  = shp_district2.to_crs(target_crs)

shp_district2.plot(figsize=(12, 12), facecolor='white', edgecolor='black', linewidth=0.5)

plt.title("District 2 map")
plt.show()
```

```python
# shape file for entire Africa in case you need that
shp_district2 = gpd.read_file("merged_adm2.shp")

target_crs = "EPSG:4326"

#combine the df
shp_Africa     = shp_Africa.to_crs(target_crs)
shp_district2  = shp_district2.to_crs(target_crs)
gdf_combined = gpd.GeoDataFrame(pd.concat([shp_Africa, shp_district2], ignore_index=True), crs=target_crs)

gdf_combined.plot(figsize=(12, 12), facecolor='white', edgecolor='black', linewidth=0.5)

plt.title("District 2 + Large Africa country map")
plt.show()
```
